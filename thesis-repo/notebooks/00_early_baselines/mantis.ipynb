```python
import numpy as np
import torch
import torch.nn.functional as F
import gc
import sys
from sklearn.linear_model import RidgeClassifier
from sklearn.metrics import accuracy_score, classification_report

sys.path.insert(0, '/kaggle/working/FM4MTSC/foundation_models/mantis')
from architecture import Mantis8M
from trainer.trainer import MantisTrainer

device = "cuda" if torch.cuda.is_available() else "cpu"
print('Device:', device)

network = Mantis8M(device=device)
network = network.from_pretrained("paris-noah/Mantis-8M")
print('Mantis loaded')

def resize(X):
    return F.interpolate(
        torch.tensor(X, dtype=torch.float),
        size=512, mode="linear", align_corners=False
    ).numpy()
```

```python
data = np.load('/kaggle/input/datasets/aislingwallace/diss-datasets-2/SYNTH.npy', allow_pickle=True).item()
X_tr = data['train']['X']
y_tr = np.array([int(x) for x in data['train']['y']])
X_te = data['test']['X']
y_te = np.array([int(x) for x in data['test']['y']])

print('Resizing...')
X_tr_r = resize(X_tr)
X_te_r = resize(X_te)

print('Extracting embeddings...')
model = MantisTrainer(device=device, network=network)
Z_tr = model.transform(X_tr_r)
Z_te = model.transform(X_te_r)

np.save('/kaggle/working/Mantis_SYNTH_train_emb.npy', Z_tr)
np.save('/kaggle/working/Mantis_SYNTH_test_emb.npy', Z_te)

clf = RidgeClassifier(random_state=42)
clf.fit(Z_tr, y_tr)
y_pred = clf.predict(Z_te)
print(f'SYNTH Accuracy: {accuracy_score(y_te, y_pred):.4f}')
print(classification_report(y_te, y_pred))
del Z_tr, Z_te, X_tr, X_te, X_tr_r, X_te_r, data
gc.collect()
torch.cuda.empty_cache()
```
