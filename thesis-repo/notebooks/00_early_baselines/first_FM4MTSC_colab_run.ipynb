```python
from momentfm import MOMENTPipeline

model = MOMENTPipeline.from_pretrained(
    "AutonLab/MOMENT-1-large",
    model_kwargs={
        "task_name": "classification",
        "n_channels": 8,
        "num_class": len(set(y_train))
    },
)
model.init()
print(model)
```

```python
import torch

X_batch = torch.tensor(X_train[:4], dtype=torch.float32)

model.eval()
with torch.no_grad():
    output = model(x_enc=X_batch)

print(output)
print(output.logits.shape)
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
y_train_enc = le.fit_transform(y_train)
y_test_enc = le.transform(y_test)

print(y_train_enc[:5])

y_tr = torch.tensor(y_train_enc, dtype=torch.long)
y_te = torch.tensor(y_test_enc, dtype=torch.long)
```

```python
from torch.utils.data import DataLoader, TensorDataset
from torch.optim import Adam
import torch.nn as nn
```

```python
X_tr = torch.tensor(X_train, dtype=torch.float32)
X_te = torch.tensor(X_test, dtype=torch.float32)

train_loader = DataLoader(TensorDataset(X_tr, y_tr), batch_size=32, shuffle=True)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(device)
model = model.to(device)

optimizer = Adam(model.head.parameters(), lr=1e-3)
criterion = nn.CrossEntropyLoss()
```

```python
model.train()
for epoch in range(5):
    total_loss = 0
    for X_batch, y_batch in train_loader:
        X_batch, y_batch = X_batch.to(device), y_batch.to(device)

        optimizer.zero_grad()
        output = model(x_enc=X_batch)
        loss = criterion(output.logits, y_batch)
        loss.backward()
        optimizer.step()
        total_loss += loss.item()

    print(f"Epoch {epoch+1} | Loss: {total_loss/len(train_loader):.4f}")
```

```python
torch.save(model.state_dict(), '/content/drive/MyDrive/Datasets/MOMENT_SYNTH_trained.pt')
print("Model saved!")
```

```python
from sklearn.metrics import accuracy_score, classification_report

model.eval()
all_preds = []

with torch.no_grad():
    for i in range(0, len(X_te), 32):
        X_batch = X_te[i:i+32].to(device)
        output = model(x_enc=X_batch)
        preds = output.logits.argmax(dim=1).cpu().numpy()
        all_preds.extend(preds)

print(accuracy_score(y_te.numpy(), all_preds))
print(classification_report(y_te.numpy(), all_preds))
```
