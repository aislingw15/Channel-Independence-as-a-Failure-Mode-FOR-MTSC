```python
import numpy as np
import torch
from sklearn.linear_model import RidgeClassifier
from sklearn.metrics import accuracy_score, classification_report
from chronos import Chronos2Pipeline

device = "cuda" if torch.cuda.is_available() else "cpu"
print('Device:', device)

pipeline = Chronos2Pipeline.from_pretrained(
    "amazon/chronos-2",
    device_map=device,
)
print('Chronos-2 loaded')

def get_chronos2_embeddings(pipeline, X, batch_size=32):
    all_embeddings = []
    for i in range(0, len(X), batch_size):
        batch = X[i:i+batch_size]
        context = torch.tensor(batch, dtype=torch.float32)
        embeddings, _ = pipeline.embed(context)
        for emb in embeddings:
            pooled = emb.mean(dim=1).flatten().numpy()
            all_embeddings.append(pooled)
        print(f'  {min(i+batch_size, len(X))}/{len(X)} done')
    return np.array(all_embeddings)
```

```python
data = np.load('/content/datasets/CMJ.npy', allow_pickle=True).item()
X_tr = data['train']['X']
y_tr = np.array([int(x) for x in data['train']['y']])
X_te = data['test']['X']
y_te = np.array([int(x) for x in data['test']['y']])

train_emb = get_chronos2_embeddings(pipeline, X_tr)
test_emb = get_chronos2_embeddings(pipeline, X_te)

clf = RidgeClassifier()
clf.fit(train_emb, y_tr)
y_pred = clf.predict(test_emb)

acc = accuracy_score(y_te, y_pred)
print(f'Chronos-2 Accuracy on CMJ: {acc:.4f}')
print(classification_report(y_te, y_pred))

np.save('/content/drive/MyDrive/Chronos2_CMJ_train_emb.npy', train_emb)
np.save('/content/drive/MyDrive/Chronos2_CMJ_test_emb.npy', test_emb)
```
