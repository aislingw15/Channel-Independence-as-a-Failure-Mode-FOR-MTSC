```python
import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch, FancyArrowPatch
import matplotlib as mpl
from pathlib import Path

FIG = Path('/content/drive/MyDrive/diss-figures'); FIG.mkdir(parents=True, exist_ok=True)
mpl.rcParams.update({'font.family':'serif','savefig.dpi':300,
                     'savefig.bbox':'tight','figure.facecolor':'white'})

INK='#2B2D42'; MUTED='#8D99AE'; TEAL='#2A9D8F'; CORAL='#E76F51'
SAND='#E9C46A'; SLATE='#556270'; CHBG='#EDF2F4'

fig, axes = plt.subplots(1, 2, figsize=(10, 4.6))
fig.patch.set_facecolor('white')

def chip(ax,x,y,w,h,text,fc,tc=INK,fs=9):
    ax.add_patch(FancyBboxPatch((x,y),w,h,boxstyle='round,pad=0.02,rounding_size=0.08',
                 linewidth=1.0,edgecolor=fc,facecolor=fc,zorder=3))
    ax.text(x+w/2,y+h/2,text,ha='center',va='center',fontsize=fs,color=tc,zorder=4)

def curve(ax,x1,y1,x2,y2,color,lw=1.6,alpha=1.0):
    ax.add_patch(FancyArrowPatch((x1,y1),(x2,y2),arrowstyle='-|>',mutation_scale=13,
                 lw=lw,color=color,alpha=alpha,zorder=2))

ch_y=[6.4,5.1,3.8]

ax=axes[0]
ax.text(4.6,8.4,'Channel-aware',ha='center',fontsize=13,color=INK,fontweight='bold')
ax.text(4.6,7.85,'channels interact as features form',ha='center',fontsize=8.5,
        color=MUTED,style='italic')
for i,y in enumerate(ch_y):
    chip(ax,0.4,y,1.5,0.85,f'ch {i+1}',CHBG,fs=8.5)
    curve(ax,1.9,y+0.42,3.2,5.25,TEAL,lw=1.5,alpha=0.85)
chip(ax,3.2,4.3,2.3,1.9,'shared\nencoder',TEAL,tc='white',fs=10)
curve(ax,5.5,5.25,7.2,5.25,SLATE,lw=1.7)
chip(ax,7.2,4.65,1.5,1.2,'classifier',SAND,fs=9)
ax.text(4.35,3.0,'the relationship\nis preserved',ha='center',fontsize=8.5,
        color=TEAL,style='italic',fontweight='bold')

ax=axes[1]
ax.text(4.9,8.4,'Channel-independent',ha='center',fontsize=13,color=INK,fontweight='bold')
ax.text(4.9,7.85,'each channel encoded in isolation',ha='center',fontsize=8.5,
        color=MUTED,style='italic')
for i,y in enumerate(ch_y):
    chip(ax,0.3,y,1.3,0.85,f'ch {i+1}',CHBG,fs=8.5)
    curve(ax,1.6,y+0.42,2.5,y+0.42,SLATE,lw=1.4)
    chip(ax,2.5,y,1.6,0.85,'encoder',MUTED,tc='white',fs=8)
    curve(ax,4.1,y+0.42,5.3,5.6,SLATE,lw=1.4,alpha=0.8)
    ax.plot([2.15,2.15],[y-0.05,y+0.9],color=CORAL,lw=0.8,ls=(0,(2,2)),alpha=0.5,zorder=1)
chip(ax,5.3,4.9,1.4,1.0,'concat',INK,tc='white',fs=8.5)
curve(ax,6.7,5.4,7.5,5.4,SLATE,lw=1.7)
chip(ax,7.5,4.9,1.5,1.0,'classifier',SAND,fs=9)
ax.text(3.2,2.9,'anything between channels\nis gone before this point',ha='center',
        fontsize=8.5,color=CORAL,style='italic',fontweight='bold')

for ax in axes:
    ax.set_xlim(0,9.2); ax.set_ylim(2.4,8.9); ax.axis('off')

plt.tight_layout()
for ext in ('pdf','png'):
    plt.savefig(FIG/f'fig1_1_channel_processing.{ext}')
plt.show()
print('saved fig1_1_channel_processing')
```

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib as mpl
from pathlib import Path

FIG = Path('/content/drive/MyDrive/diss-figures')
mpl.rcParams.update({'font.family': 'serif', 'savefig.dpi': 300,
                     'savefig.bbox': 'tight', 'figure.facecolor': 'white'})
BLUE='#0072B2'; ORANGE='#E69F00'; GREY='#555555'

t = np.linspace(0, 4*np.pi, 300)
rng = np.random.default_rng(0)

fig, axes = plt.subplots(2, 2, figsize=(8.4, 4.4), sharex=True)

def plot_pair(col, title, cA, cB, note):
    for row, (sig, lab) in enumerate([(cA, 'channel 1'), (cB, 'channel 2')]):
        ax = axes[row][col]
        ax.plot(t, sig, color=BLUE if col == 0 else GREY, lw=1.1)
        ax.set_ylabel(lab, fontsize=8)
        ax.set_yticks([]); ax.set_xticks([])
        for s in ax.spines.values(): s.set_visible(False)
    axes[0][col].set_title(title, fontsize=10, pad=8)
    axes[1][col].annotate(note, xy=(0.5, -0.35), xycoords='axes fraction',
                          ha='center', fontsize=8, style='italic',
                          color=BLUE if col == 0 else GREY)

within_A = np.sin(t) + 0.08*rng.standard_normal(len(t))
within_B = 0.15*rng.standard_normal(len(t))
plot_pair(0, 'Within-channel signal',
          within_A, within_B,
          'the label is visible\nin channel 1 alone')

base = rng.standard_normal(len(t)) * 0.5
between_A = base + 0.1*rng.standard_normal(len(t))
between_B = base + 0.1*rng.standard_normal(len(t))
plot_pair(1, 'Between-channel signal',
          between_A, between_B,
          'neither channel alone is informative;\nthe label is in their correlation')

plt.tight_layout()
for ext in ('pdf', 'png'):
    plt.savefig(FIG / f'fig1_2_channel_dependency.{ext}')
plt.show()
print('saved fig1_2_channel_dependency')
```
